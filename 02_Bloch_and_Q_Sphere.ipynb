# Bloch Sphere and Q-Sphere Visualization

This experiment explores the visualization of a single-qubit quantum state using the **Bloch sphere** and **Q-sphere** representations.

Starting from the initial state $|0\rangle$, different quantum gates are applied sequentially to observe how they transform the qubit state.

The experiment demonstrates the effect of the **X, H, and Z gates** and compares the resulting quantum states using their statevectors and visual representations.

```python
import qiskit
```

```python
from qiskit import QuantumCircuit
from qiskit.visualization import plot_state_qsphere
from qiskit.visualization import plot_bloch_multivector
```

## Creating a Single Qubit

A quantum circuit containing a single qubit is initialized.

By default, a qubit in Qiskit starts in the computational basis state $|0\rangle$.

We will use this qubit to observe how different quantum gates transform its state.

```python
qc = QuantumCircuit(1)
```

## Initial State $|0\rangle$

Before applying any quantum gates, the qubit is in the state

$$
|0\rangle =
\begin{bmatrix}
1 \\
0
\end{bmatrix}.
$$

The statevector provides the mathematical representation of the quantum state, while the Bloch sphere and Q-sphere provide visual representations of the same state.

```python
from qiskit.quantum_info import Statevector

statevector = Statevector.from_instruction(qc)

statevector
```

```python
plot_bloch_multivector(statevector)
```

```python
plot_state_qsphere(statevector)
```

## Applying the X Gate

The **Pauli-X gate** acts as a quantum bit-flip operation.

It transforms the computational basis states according to

$$
X|0\rangle = |1\rangle
$$

and

$$
X|1\rangle = |0\rangle.
$$

Applying the X gate to our initial qubit therefore moves the state from $|0\rangle$ to $|1\rangle$.

```python
qc.x(0)
```

```python
statevector = Statevector.from_instruction(qc)

statevector
```

```python
plot_bloch_multivector(statevector)
```

```python
plot_state_qsphere(statevector)
```

## Applying the Hadamard Gate

The **Hadamard (H) gate** creates a superposition of the computational basis states.

When applied to $|1\rangle$,

$$
H|1\rangle =
\frac{|0\rangle - |1\rangle}{\sqrt{2}}.
$$

This state is commonly denoted as

$$
|-\rangle =
\frac{|0\rangle - |1\rangle}{\sqrt{2}}.
$$

The Bloch sphere provides an intuitive geometric representation of this transformation.

```python
qc.h(0)
```

```python
statevector = Statevector.from_instruction(qc)

statevector
```

```python
plot_bloch_multivector(statevector)
```

```python
plot_state_qsphere(statevector)
```

## Applying the Z Gate

The **Pauli-Z gate** introduces a phase flip by changing the sign of the $|1\rangle$ component:

$$
Z|0\rangle = |0\rangle
$$

and

$$
Z|1\rangle = -|1\rangle.
$$

Applying the Z gate to the state

$$
|-\rangle =
\frac{|0\rangle - |1\rangle}{\sqrt{2}}
$$

gives

$$
Z|-\rangle =
\frac{|0\rangle + |1\rangle}{\sqrt{2}}
= |+\rangle.
$$

Thus, the Z gate changes the relative phase between the two components of the superposition.

```python
qc.z(0)
```

```python
statevector = Statevector.from_instruction(qc)

statevector
```

```python
plot_state_qsphere(statevector)
```

## Observations

The sequence of gates demonstrates how quantum operations modify a qubit state:

- The **X gate** transforms $|0\rangle$ into $|1\rangle$.
- The **H gate** creates the superposition state $|-\rangle$.
- The **Z gate** changes the relative phase of the superposition, transforming $|-\rangle$ into $|+\rangle$.
- The **Bloch sphere** provides a geometric representation of the qubit state on the surface of the sphere.
- The **Q-sphere** represents the probability amplitudes and phases associated with the computational basis states.
